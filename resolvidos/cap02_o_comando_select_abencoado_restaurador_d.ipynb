## A diferença entre AND e OR · PDF p. 116

#### Exercício: Parênteses mudam o resultado

Compare, contando as linhas com `COUNT(*)`: `sexo = 'F' AND local = 'Recife, PE' OR profissao = 'Chef'` contra `sexo = 'F' AND (local = 'Recife, PE' OR profissao = 'Chef')`. O `AND` é avaliado antes do `OR`.

```python
%%sql
SELECT
  (SELECT COUNT(*) FROM meus_contatos WHERE sexo = 'F' AND local = 'Recife, PE' OR profissao = 'Chef') AS sem_parenteses,
  (SELECT COUNT(*) FROM meus_contatos WHERE sexo = 'F' AND (local = 'Recife, PE' OR profissao = 'Chef')) AS com_parenteses;
```

## Usando IS NULL para encontrar NULL · PDF p. 118

#### Exercício: IS NULL (e não = NULL)

Mostre quem está sem email e quem está sem estado civil. Depois confira que `WHERE email = NULL` não traz nada.

```python
%%sql
SELECT primeiro_nome, email, estado_civil
FROM meus_contatos
WHERE email IS NULL OR estado_civil IS NULL;
```

## Bem BETWEEN (entre) nós... há um jeito melhor · PDF p. 123

#### Exercício: BETWEEN

Mostre as notas de 7 a 9 (inclusive) e as visitas entre `2024-03-05` e `2024-03-10` (aqui, as duas condições juntas com AND).

```python
%%sql
SELECT * FROM notas_cafeterias
WHERE nota BETWEEN 7 AND 9 AND visita BETWEEN '2024-03-05' AND '2024-03-10';
```

## Depois dos encontros, você está IN... · PDF p. 125

#### Exercício: IN

Mostre os contatos que moram em Recife, Olinda ou Natal usando `IN`.

```python
%%sql
SELECT primeiro_nome, local FROM meus_contatos
WHERE local IN ('Recife, PE', 'Olinda, PE', 'Natal, RN');
```

## ...ou você está NOT IN · PDF p. 125

#### Exercício: NOT IN

Agora os que **não** moram nessas três cidades.

```python
%%sql
SELECT primeiro_nome, local FROM meus_contatos
WHERE local NOT IN ('Recife, PE', 'Olinda, PE', 'Natal, RN');
```

## Mais NOT · PDF p. 126

#### Exercício: NOT com BETWEEN, LIKE e NULL

Mostre as bebidas cujo `qtd1` **não** está entre 3 e 4, cujo nome **não** começa com `B` e que têm instruções preenchidas (`IS NOT NULL`).

```python
%%sql
SELECT nome_bebida, qtd1 FROM bebidas_faceis
WHERE qtd1 NOT BETWEEN 3 AND 4
  AND nome_bebida NOT LIKE 'B%'
  AND instrucoes IS NOT NULL;
```
