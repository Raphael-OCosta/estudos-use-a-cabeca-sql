## Os dois passos do INSERT e DELETE · PDF p. 147

#### Exercício: Inserir o novo, depois apagar o velho

O `Tico-Tico` mudou de lugar: foi visto no `Parque da Jaqueira`. Faça em dois passos: insira a linha nova e apague a antiga (`Shopping Boa Vista`). Por que inserir **antes** de apagar?

## Cuidado com o seu DELETE · PDF p. 152

#### Exercício: Teste antes com SELECT (e com ROLLBACK)

Um `DELETE` sem `WHERE` apaga **tudo**. Veja isso com segurança: `BEGIN;` abre uma transação, o `DELETE` apaga, `ROLLBACK;` desfaz. Termine contando as linhas para confirmar que nada foi perdido. (Transações são o assunto do capítulo 11.)

```python
%%sql
BEGIN;
DELETE FROM palhacos_info;
ROLLBACK;
SELECT COUNT(*) AS linhas_continuam_la FROM palhacos_info;
```

## Altere seus dados com UPDATE · PDF p. 156

#### Exercício: UPDATE em uma coluna

Troque o `UPDATE` pelos dois passos do exercício anterior: a `Lulu` foi vista no `Teatro Santa Isabel`.

```python
%%sql
UPDATE palhacos_info SET ultima_vez_visto = 'Teatro Santa Isabel' WHERE nome = 'Lulu';
SELECT * FROM palhacos_info WHERE nome = 'Lulu';
```

## UPDATE em ação · PDF p. 159

#### Exercício: UPDATE em várias colunas

O `Fuxico` foi visto na `Feira de Casa Amarela` e agora também faz `mágica`. Atualize as duas colunas num único `UPDATE`.

```python
%%sql
UPDATE palhacos_info
SET ultima_vez_visto = 'Feira de Casa Amarela',
    atividades = 'malabarismo, buzina, mágica'
WHERE nome = 'Fuxico';
SELECT * FROM palhacos_info WHERE nome = 'Fuxico';
```

## UPDATE (atualize) seus preços · PDF p. 164

#### Exercício: Preço com expressão

Aumente em R$ 1,00 o preço das bebidas **com gelo** (`gelo = 'S'`) usando `preco = preco + 1`.

```python
%%sql
UPDATE bebidas_precos SET preco = preco + 1 WHERE gelo = 'S';
SELECT nome_bebida, preco, gelo FROM bebidas_precos ORDER BY nome_bebida;
```

## Tudo o que precisamos é um UPDATE · PDF p. 166

#### Exercício: UPDATE sem WHERE, de propósito

Todas as bebidas ganharam uma versão sem açúcar com 20% menos calorias. Atualize a coluna `calorias` de **todas** as linhas (`calorias * 0.8`, arredondado com `ROUND`).

```python
%%sql
UPDATE bebidas_precos SET calorias = ROUND(calorias * 0.8);
SELECT nome_bebida, calorias FROM bebidas_precos;
```
